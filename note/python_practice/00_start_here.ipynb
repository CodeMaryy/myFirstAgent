## 第 3 格：条件判断 `if`

程序开始"动脑子"：满足条件才做某事。把 `score` 改成别的数字试试。

```python
score = 85

if score >= 90:
    print("优秀！")
elif score >= 60:
    print("及格啦")
else:
    print("要加油！")

# 注意：Python 用【缩进】(4个空格) 表示"这些代码属于 if"
```

## 第 4 格：循环 `for`

让同一件事重复做。Agent 处理多条消息、多个工具，全靠它。

```python
tools = ["搜索", "天气", "计算器", "代码执行"]

# for index, tool in enumerate(tools):
#     print(f"工具 {index + 1}: {tool}")

# enumerate 会自动给每个元素编号（从 0 开始）

# 写法 1：只要值 —— 最常用
for tool in tools:
    print(tool)

# 写法 2：只要编号 —— range + len（就是 Java 的 for i 循环）
for i in range(len(tools)):
    print(tools[i])

# 写法 3：两个都要 —— enumerate + 解包 ★最 Pythonic
for index, tool in enumerate(tools):
    print(index, tool)
```

## 第 5 格：函数 `def`

把一段逻辑打包，起个名字，随时调用。这是 Agent "工具" 的本质。

```python
def ask_llm(question: str) -> str:
    """假装这是调用 LLM（先不联网，返回固定答案）"""
    return f"你问的是：{question}，记下了！"

# 调用它
answer = ask_llm("报销流程是什么？")
print(answer)
print(ask_llm("怎么学 Python？"))
```

## 第 6 格：类 `class`（复习 self）

你问过的 `self`：**方法第一个参数，自动接收"谁在调用我"**。

运行后观察：两个机器人互相认识，但各自的 `energy` 是独立的。

```python
class Robot:
    def __init__(self, name: str, energy: int):
        self.name = name      # self = "我自己"，把名字存到"我自己"身上
        self.energy = energy

    def work(self, hours: int):
        self.energy -= hours   # 消耗的是"我这个实例"的电量
        print(f"{self.name} 工作了 {hours} 小时，剩余电量 {self.energy}")

robot_a = Robot("小美", 100)
robot_b = Robot("小码", 100)

robot_a.work(30)
robot_b.work(10)

print(f"小美还剩 {robot_a.energy}，小码还剩 {robot_b.energy}，互不影响")
```
