# 练习 2：yield 生成器与流式输出

**AI 应用场景**

- LLM 流式输出：ChatGPT "一个字一个字往外蹦" 的背后就是 yield
- 读取大文件/大向量库：一次取一点，不撑爆内存

**核心语法**

- `yield x`：暂停函数并吐出一个值，函数状态被冻结，下次从暂停处继续
- 普通函数 + `return`：一次性算完返回整个列表（等得久、占内存）
- 生成器 + `yield`：**惰性产出**，用到才算、算到哪取到哪

```python
from typing import Generator

def fake_llm_stream(text: str) -> Generator[str, None, None]:
    # 模拟 LLM 流式返回：每次 yield 一个词
    for token in text.split(" "):
        yield token  # 暂停！交出控制权，调用方拿到这个值

print("fake_llm_stream 已定义 ✅（注意：定义时里面的代码一行都没执行）")
```

```python
# 拿到一个打印一个，不用等全部完成 —— 打字机效果
print("🤖 LLM:", end=" ", flush=True)
for token in fake_llm_stream("你好 我是 流式 输出 的 LLM 一个 字 一个 字 返回"):
    print(token, end=" ", flush=True)
print()
```

```python
# 对比：直接转成 list = 必须等函数全部执行完才能拿到
tokens = list(fake_llm_stream("一次性 返回 所有 内容"))
print(f"拿到的列表: {tokens}")
```

## 💡 小结

| | 流式（yield） | 一次性（list） |
|---|---|---|
| 第一个字出现时间 | **立刻** | 等全部生成完 |
| 内存占用 | 一个 token | 整段文本 |
| 体验 | 打字机效果 🎉 | 用户干等 |
